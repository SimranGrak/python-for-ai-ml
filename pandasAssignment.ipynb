```python
# Pandas Assignment
''' Start by loading the IRIS dataset.
A. Display the following information:
i. first 10 rows
ii. shape and datatypes
iii. summary statistics(mean, std, min, max)
'''

import pandas as pd
df=pd.read_csv("IRIS.csv")
df.head(10)
df.shape
df.dtypes

df[["sepal_length", "sepal_width", "petal_length", "petal_width"]].agg(
    ["mean", "std", "min", "max"]
)


'''select only those rows where:
i. petal_length> 4.5
ii. species="Iris-virginica"
'''

result=df[(df["petal_length"]> 4.5) & (df["species"]=="Iris-virginica")]
result, result.shape



'''Group by species and compute
i. average sepal_length
ii. maxinum petal_width
iii. standard deviation of sepal_width
'''

group_result=df.groupby("species").agg(
    avg_sepal_length=("sepal_length", "mean"),
    max_petal_width=("petal_width", "max"),
    std_sepal_width=("sepal_width", "std")
)

group_result


'''create a new column: "petal_ratio"= petal_length/petal_width
For each species, find average petal_ratio 
'''

new_column=df.assign(petal_ratio= df["petal_length"]/df["petal_width"])

new_column.groupby("species")["petal_ratio"].mean()
```

```python
'''start by loading  the TITANIC dataset.
A. display only columns: Name, Sex, Age, Fare, Survived
'''

df=pd.read_csv("Titanic-Dataset.csv")
df[["Name", "Sex", "Age", "Fare", "Survived"]]


'''Select passengers who:
i. are females
ii. Fare> 30
'''

df[(df["Sex"]=="female") & (df["Fare"]>30)]


'''Groupby Pclass and compute:
i. Survival rate
ii. Average Fare
iii. Average age
'''

df.groupby("Pclass").agg(
    survival_rate=("Survived","mean"),
    avg_fare=("Fare", "mean"),
    avg_age=("Age", "mean")
)
```
