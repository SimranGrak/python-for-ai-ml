27. You have marks of 5 students in 3 subjects:
marks = np.array([
    [80, 75, 90],
    [60, 65, 70],
    [95, 90, 85],
    [50, 55, 60],
    [88, 92, 90]
])

Find:

average marks of each student
average marks of each subject
highest mark
lowest mark
student(s) whose average is above 80

```python
marks = np.array([
    [80, 75, 90],
    [60, 65, 70],
    [95, 90, 85],
    [50, 55, 60],
    [88, 92, 90]
])

print("average marks of each student:", np.mean(marks, axis=1) )
print("average marks of each subject:", np.mean(marks, axis=0) )
print("highest marks:", np.max(marks) )
print("lowest marks:", np.min(marks) )
average_marks=np.mean(marks, axis=1)
print("marks above 80:", (marks[average_marks > 80]))
```
