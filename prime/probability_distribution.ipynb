# Normal distribution

```python
from scipy.stats import norm

sigma=10   #standard deviation
mu=70      #mean

#x-axis values
x=np.linspace(30, 110, 1000)

#normal pdf
y=norm.pdf(x, mu, sigma)

#plot a bell curve
plt.plot(x, y, color="black", linewidth=2, label="normal distribution")

#shade 68% region(mu +-1sigma)
x1=np.linspace(mu-sigma, mu+sigma, 1000)
plt.fill_between(x1, norm.pdf(x1, mu, sigma), color="green",alpha=0.3, linewidth=2, label="68% region")

#shade 95% region(mu +-2sigma)
x2=np.linspace(mu-2*sigma, mu+2*sigma, 1000)
plt.fill_between(x2, norm.pdf(x2, mu, sigma), color="yellow",alpha=0.3, linewidth=2, label="95% region")

#shade 99.7% region(mu +-3sigma)
x3=np.linspace(mu-3*sigma, mu+3*sigma, 1000)
plt.fill_between(x3, norm.pdf(x3, mu, sigma), color="red", alpha=0.3, linewidth=2, label="99.7%")

plt.title("noraml distribution of student scores")

plt.legend()
```
