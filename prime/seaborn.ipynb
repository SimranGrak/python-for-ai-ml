```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
print(sns.get_dataset_names())
```

```python
tips=sns.load_dataset("tips")
tips.head()
```

```python
sns.set_theme(style="ticks")

sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    size="size",
    style="sex"
)
```

```python
# scatter plot

tips=sns.load_dataset("tips")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time"
)
```

```python
sns.lineplot(
    data=tips,
    x="total_bill",
    y="tip",
)
```

```python
#barplot

sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex"
)
```

```python
#box plot

sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
)
```

```python
#histrogram

penguins=sns.load_dataset("penguins")

sns.histplot(
    data=penguins,
    x="body_mass_g",
    bins=30
)
```

```python
#heatmap

flights=sns.load_dataset("flights")

flights.head()
```

```python
import pandas as pd

flights_pivot=flights.pivot(index="month", columns="year", values="passengers")

flights_pivot.head()

sns.heatmap(
    flights_pivot,
    cmap="coolwarm",
    annot=True,
    fmt="d"
)

plt.title("passengers heatmap")
plt.tight_layout()
```
