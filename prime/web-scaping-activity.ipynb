```python
import requests
from bs4 import BeautifulSoup
```

```python
#extract data from all pages

page_count=1

while True:
    URL=f"https://quotes.toscrape.com/page/{page_count}/"
    res=requests.get(URL)

    soup=BeautifulSoup(res.text, "lxml")
    quotes=soup.select("div.quote")

    if not quotes:
        print("no more valid pages...")
        break

    with open(f"scraped_data/quotes{page_count}.html", "w", encoding="UTF-8") as f:
        f.write(res.text)
        print(f"downloaded data from page{page_count}")

    page_count+=1
```

```python
#extract useful information

import glob

all_quotes=[]

for i in glob.glob("scraped_data/quotes*.html"):
    with open(i, "r", encoding="UTF-8") as f:
        html_content=f.read()

    soup=BeautifulSoup(html_content, "lxml")

    all_quotes=soup.select("div.quote")

    for q in all_quotes:
        all_tags=[]

        for tag in q.select(".tags .tag"):
            all_tags.append(tag.get_text())

        if 'life' in all_tags:
            text=q.select_one("span.text").get_text()
            author=q.select_one("small.author").get_text()
            life_quotes.append([text, author])
```

```python
import pandas as pd

df=pd.DataFrame(life_quotes, columns=["text", "quotes"])

df.to_csv("cleaned_data/quotes-data.csv", index=False)
```
