1. For the following dataset generated using NumPy:
   np.random.normal(loc=70, scale=10, size=200)
   •Plot a of scores histogram
   •Draw a vertical line for the mean & try to interpret the results.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
np.random.seed(0)

data=np.random.normal(70, 10, 200)

plt.hist(data, bins=15, edgecolor="black")

average=np.mean(data)

plt.axvline(
    x=average,
    label="average marks",
    linewidth=2,
    linestyle="--",
    color="red"
)

plt.title("number frequency")
plt.xlabel("numbers range")
plt.ylabel("frequency")
plt.tight_layout()
```

3. For the Seaborn tips dataset, do the following:
   •Create a box plot of total_bill for each day
   •add hue "sex
   •Identify which day has the highest median bill
   •Comment on outliers

```python
tips=sns.load_dataset("tips")

sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex"
)

highest_median=tips.groupby("day")["total_bill"].median()
print(highest_median)

print("day with maximum median:", highest_median.idxmax())
```

4. For the Seaborn tips dataset, do the following:
   •Create a scatter plot of total_bill vs tip
   •Color points by sex
   •Use different markers for smoker
   •Add a title and axis labels
   •Analyze:Do smokers tend to tip differently?

```python
tips=sns.load_dataset("tips")

sns.relplot(
    data=tips,
    x="tip",
    y="total_bill",
    hue="smoker",
    style="sex"
)

plt.title("total_bill vs tips")
```

5. For the Seaborn tips dataset, do the following:
   •Create a pivot table of average tip
       ◦Rows:day
       ◦Columns:time
   •Plot the pivot table as a heatmap
   •Annotate values
   •Interpret which day–time combination performs best

```python
tips=sns.load_dataset("tips")

pivot_table=tips.pivot_table(index="day", columns="time", values="tip", aggfunc="mean")

sns.heatmap(
    pivot_table,
    cmap="coolwarm",
    annot=True,
    fmt=".2f"
)
```
