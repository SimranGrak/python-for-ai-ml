```python
# count minimum value of quantity

print(df.groupby("InvoiceNo")["Quantity"].min())

print(df["Quantity"].agg(["min", "max", "mean","std"]))

#return row that contains minimum and maximum value because it will outlier for us
print(df.loc[df["Quantity"].agg(["idxmin", "idxmax"])])
```

```python
#how many rows in dataset contains InvoiceNo starts with c and how many rows with c

cancel_transactions=[]
non_cancel_transactions=[]

df["InvoiceNo"]=df["InvoiceNo"].astype(str)

for i in df["InvoiceNo"]:
    if "C" in i:
        cancel_transactions.append(i)
    else:
        non_cancel_transactions.append(i)

cancel_len=len(cancel_transactions)
non_cancel_len=len(non_cancel_transactions)
total_rows=len(df)

print("percentage of cancel transactions:", (cancel_len/total_rows)*100, "%")
print("percentage of non-cancel transactions:", (non_cancel_len/total_rows)*100, "%")
```

```python
# new dataframe for cancel and non_cancel transactions

df["InvoiceNo"]=df["InvoiceNo"].astype(str)

non_cancel_df=df[df["InvoiceNo"].str.fullmatch(r"\d+")].copy()

cancel_df=df[df["InvoiceNo"].str.startswith("C")].copy()
non_cancel_df
```

```python
negative_quantity=cancel_df[cancel_df["Quantity"] < 0]

negative_quantity
```

```python
#EDA on non_cancel_df

non_cancel_df["Quantity"].agg(["min", "max"])

print(non_cancel_df.loc[non_cancel_df["Quantity"].agg(["idxmin", "idxmax"])])
```

```python
# look at the negative Quantity rows of non-cancel_df

negative_quantity=non_cancel_df[non_cancel_df["Quantity"]< 0]
negative_quantity

#look at the positive quantity rows of non-cancel_df
positive_quantity=non_cancel_df[non_cancel_df["Quantity"]>= 0]
positive_quantity

#new dataframe of negative quantity for EDA
neg_quan_df=non_cancel_df.loc[non_cancel_df["Quantity"] < 0].copy()

#check customer_id in neg_quan_df
check_customerID=neg_quan_df[neg_quan_df["CustomerID"].isnull()]
check_customerID
```
